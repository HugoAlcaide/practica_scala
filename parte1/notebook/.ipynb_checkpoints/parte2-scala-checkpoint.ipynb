## Ejercicio 1 — Variables, tipos e inferencia

**Explicación:** 
En este ejercicio se definen variables para representar la información de una partida de cartas empleando dos enfoques distintos:
* **Versión A:** Se declaran las variables indicando de forma explícita el tipo de dato correspondiente (`String`, `Int`, `Double`, `Float`, `Boolean`).
* **Versión B:** Se declaran las mismas variables omitiendo el tipo de dato, permitiendo que el compilador de Scala realice la inferencia de tipos de manera automática a partir de los valores asignados.

// Versión A: Declaración explícita de tipos
val nombreA: String = "Hugo"
val partidasA: Int = 12
val puntuacionA: Double = 18.5
val porcentajeVictoriasA: Float = 0.75f
val activoA: Boolean = true

// Versión B: Inferencia de tipos por parte de Scala
val nombreB = "Hugo"
val partidasB = 12
val puntuacionB = 18.5
val porcentajeVictoriasB = 0.75
val activoB = true

**Conclusiones y Documentación:**
* **Tipos inferidos:** Scala infiere automáticamente `String` para cadenas de texto, `Int` para valores enteros, `Double` por defecto para números con decimales (como se observa en `puntuacionB` y `porcentajeVictoriasB`), y `Boolean` para valores lógicos de verdad.
* **Diferencia entre versiones:** A nivel de rendimiento y tipado estricto final no hay diferencia, ya que Scala es un lenguaje fuertemente estático. La diferencia principal radica en la concisión y legibilidad del código: la Versión A aporta control explícito y la Versión B reduce la verbosidad mediante inferencia.
* **Val vs Var:** En este caso se han empleado variables de tipo `val` (inmutables) porque representan información estática de la partida. Si el estado del jugador tuviera que cambiar de forma dinámica durante la ejecución del programa (por ejemplo, incrementar la puntuación), se debería emplear `var`.

## Ejercicio 2 — val, var y reasignación

**Explicación:** 
En este ejercicio se crean una variable inmutable (`val`) para el nombre del jugador y una variable mutable (`var`) para la puntuación. Se realizan operaciones de incremento sobre la puntuación y posteriormente se intenta reasignar un nuevo valor al nombre del jugador para evidenciar el error de compilación.

```scala
val jugador = "Marta"
var puntuacion = 10
puntuacion = puntuacion + 5
puntuacion = puntuacion + 3
println(puntuacion)
```

```scala
jugador = "Hugo"
```

**Conclusiones y Documentación:**
* **Por qué `puntuacion` puede modificarse:** Se ha declarado como una variable de tipo `var`, lo que indica que su referencia y contenido son mutables en el tiempo y permiten reasignaciones posteriores.
* **Por qué `jugador` no puede reasignarse:** Se ha declarado mediante `val`, lo que define una referencia inmutable. Una vez asignado el texto inicial, no es posible cambiar su valor.
* **Mensaje de error generado por Scala:** Al intentar ejecutar la línea `jugador = "Hugo"`, el compilador lanza el error **`reassignment to val`**, protegiendo la inmutabilidad de la variable y bloqueando la ejecución. Tal como se requiere, este bloque de error se mantiene documentado en el Notebook.

## Ejercicio 4 — Función para determinar si una mano se pasa de 21

**Explicación:** 
Se define una función que recibe un valor entero representativo de la puntuación y evalúa mediante una expresión relacional si este supera el límite establecido (21), devolviendo un valor de tipo `Boolean`. Posteriormente se comprueba su funcionamiento con los valores de prueba indicados.

```scala
def bust(puntuacion: Int): Boolean = {
    puntuacion > 21
}

println(bust(18))
println(bust(21))
println(bust(22))
println(bust(30))
```

**Conclusiones y Documentación:**
* **Pureza de la función:** La función cumple con el requisito de ser una función pura: no modifica ninguna variable externa ni genera efectos secundarios, limitándose a evaluar el parámetro de entrada y devolver el resultado booleano correspondiente.
* **Estilo funcional:** Se ha prescindido del uso de bloques `if / else` explícitos (`if (puntuacion > 21) true else false`), simplificando la lógica directamente a la expresión booleana `puntuacion > 21`, lo cual resulta más conciso y idiomático en Scala.

## Ejercicio 5 — Comparación de dos manos

**Explicación:** 
Se define la función `maxHand` utilizando una estructura condicional de tipo `if / else if / else` para comparar dos valores enteros correspondientes a dos manos (`handA` y `handB`). Se ha modificado el comportamiento para el caso de empate (cuando ambas manos tienen el mismo valor), haciendo que la función devuelva de forma explícita el valor `1` como indicador de igualdad.

```scala
def maxHand(handA: Int, handB: Int): Int = {
  if (handA > handB) {
    handA
  } else if (handB > handA) {
    handB
  } else {
    1 // Retornar 1 en caso de empate
  }
}

println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

**Conclusiones y Documentación:**
* **Pruebas realizadas:** 
  * `maxHand(17, 19)` devuelve `19` (gana `handB`).
  * `maxHand(20, 18)` devuelve `20` (gana `handA`).
  * `maxHand(21, 21)` devuelve `1` (se detecta el empate y se aplica la modificación de retornar `1`).
* **Gestión del empate:** Cuando `handA` y `handB` tienen exactamente la misma puntuación, ninguna de las condiciones de mayor estricto (`>`) se cumple. Al utilizar el bloque `else`, se captura la igualdad y se devuelve el código `1`, evitando así elegir de forma arbitraria una mano sobre la otra.

## Ejercicio 6 — Decidir el ganador de una partida

**Explicación:** 
Se define la función `ganador` que recibe dos manos (`handA` y `handB`) y aplica las reglas del juego utilizando la función `bust` para comprobar si se superan los 21 puntos y la función `maxHand` para calcular la puntuación más alta cuando ninguna mano se pasa. Se realizan las pruebas obligatorias verificando los distintos escenarios posibles.

```scala
def ganador(handA: Int, handB: Int): Int = {
    if(bust(handA) && bust(handB)){
        0
    } else if(bust(handA)){
        handB
    } else if(bust(handB)){
        handA
    } else {
        maxHand(handA,handB)
    }
}

println(ganador(26,20))
println(ganador(18,22))
println(ganador(24,25))
println(ganador(17,19))
println(ganador(21,20))
```

**Conclusiones y Documentación de Pruebas:**
* `ganador(26, 20)` da **20**: Se cumple la condición del primer `else if` (solamente `handA` supera 21, ya que 26 > 21), por lo que gana directamente `handB` (20).
* `ganador(18, 22)` da **18**: Se cumple la condición del segundo `else if` (solamente `handB` supera 21, ya que 22 > 21), por lo que gana directamente `handA` (18).
* `ganador(24, 25)` da **0**: Se cumple la condición inicial del `if` principal (`bust(handA) && bust(handB)`), ya que ambas manos superan 21, por lo que se devuelve `0` como valor de empate por sobrepasar el límite.
* `ganador(17, 19)` da **19**: Se ejecuta el bloque `else` porque ninguna se pasa de 21; se delega en `maxHand(17, 19)` que determina que 19 es mayor.
* `ganador(21, 20)` da **21**: Se ejecuta el bloque `else` al ser ambas menores o iguales a 21; la función `maxHand(21, 20)` selecciona correctamente el 21.

## Ejercicio 7 — Arrays y mutabilidad

**Explicación:** 
En este ejercicio se crea un array mutable de cadenas de texto (`Array[String]`). Se muestra el array original, se modifica el contenido de su primera posición sustituyendo "Alex" por "Sindhu" (demostrando que el contenido del array puede cambiar a pesar de haberse declarado con `val`), y finalmente se intenta asignar un valor entero (`500`) a una de sus posiciones para comprobar las restricciones del tipado estático.

```scala
val jugadores = Array("Alex", "Chen", "Marta")
println(jugadores.mkString(", "))
jugadores(0) = "Sindhu"
println(jugadores.mkString(", "))
```

```scala
jugadores(1) = 500
```

**Conclusiones y Documentación:**
* **Mutabilidad del contenido:** Los elementos de un `Array` pueden modificarse en cualquier momento (ej. `jugadores(0) = "Sindhu"`) aunque la variable se haya declarado con `val`. El modificador `val` únicamente protege la referencia de la variable para que no pueda ser reasignada a otro array distinto, pero no congela el interior de la estructura.
* **Error de tipos (`type mismatch`):** Al intentar ejecutar `jugadores(1) = 500`, el compilador genera un error porque el array fue inferido estrictamente como `Array[String]`, por lo que rechaza introducir un valor de tipo `Int`.
* **Diferencia clave:** Reasignar la variable implicaría cambiar a qué objeto apunta (bloqueado por `val`), mientras que modificar un elemento del array altera el estado interno del objeto existente (permitido por ser una estructura mutable). Este bloque de error se mantiene documentado en el Notebook.

## Ejercicio 8 — Creación e inicialización de Arrays

**Explicación:** 
En este ejercicio se crea un array de enteros de tamaño fijo (`new Array[Int](4)`) para almacenar las puntuaciones de cuatro jugadores. Se muestra su estado inicial recién creado, se asignan de forma manual las puntuaciones correspondientes (`17, 24, 21, 19`) a cada posición, se vuelve a mostrar el array completo y se obtiene su longitud utilizando la propiedad `.length`.

```scala
val puntuaciones :Array[Int] = new Array[Int](4)
println(puntuaciones.mkString(", "))

puntuaciones(0) = 17
puntuaciones(1) = 24
puntuaciones(2) = 21
puntuaciones(3) = 19
println(puntuaciones.mkString(", "))

println(puntuaciones.length)
```

**Conclusiones y Documentación:**
* **Valores iniciales de un `Array[Int]`:** Tal como se aprecia en la primera salida (`0, 0, 0, 0`), un array de tipo numérico entero recién creado en Scala se inicializa automáticamente con ceros en todas sus posiciones antes de que se les asigne un valor de forma manual.
* **Uso de `.length`:** La propiedad `.length` permite consultar de manera dinámica el número total de elementos o la capacidad máxima establecida para el array, devolviendo en este caso `4`.
