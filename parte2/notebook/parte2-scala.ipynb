* **Versión B:** Se declaran las mismas variables omitiendo el tipo de dato, permitiendo que el compilador de Scala realice la inferencia de tipos de manera automática a partir de los valores asignados.

```scala
// Versión A: Declaración explícita de tipos
val nombreA: String = "Hugo"
val partidasA: Int = 12
val puntuacionA: Double = 18.5
val porcentajeVictoriasA: Float = 0.75f
val activoA: Boolean = true

// Versión B: Inferencia de tipos por parte de Scala
val nombreB = "Hugo"
val partidasB = 12
val puntuacionB = 18.5
val porcentajeVictoriasB = 0.75
val activoB = true
```

## Ejercicio 5 — Comparación de dos manos

**Explicación:** 
Se define la función `maxHand` utilizando una estructura condicional de tipo `if / else if / else` para comparar dos valores enteros correspondientes a dos manos (`handA` y `handB`). Se ha modificado el comportamiento para el caso de empate (cuando ambas manos tienen el mismo valor), haciendo que la función devuelva de forma explícita el valor `1` como indicador de igualdad.

```scala
def maxHand(handA: Int, handB: Int): Int = {
  if (handA > handB) {
    handA
  } else if (handB > handA) {
    handB
  } else {
    1 // Retornar 1 en caso de empate
  }
}

println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

**Conclusiones y Documentación:**
* **Pruebas realizadas:** 
  * `maxHand(17, 19)` devuelve `19` (gana `handB`).
  * `maxHand(20, 18)` devuelve `20` (gana `handA`).
  * `maxHand(21, 21)` devuelve `1` (se detecta el empate y se aplica la modificación de retornar `1`).
* **Gestión del empate:** Cuando `handA` y `handB` tienen exactamente la misma puntuación, ninguna de las condiciones de mayor estricto (`>`) se cumple. Al utilizar el bloque `else`, se captura la igualdad y se devuelve el código `1`, evitando así elegir de forma arbitraria una mano sobre la otra.

## Ejercicio 6 — Decidir el ganador de una partida

**Explicación:** 
Se define la función `ganador` que recibe dos manos (`handA` y `handB`) y aplica las reglas del juego utilizando la función `bust` para comprobar si se superan los 21 puntos y la función `maxHand` para calcular la puntuación más alta cuando ninguna mano se pasa. Se realizan las pruebas obligatorias verificando los distintos escenarios posibles.

```scala
def ganador(handA: Int, handB: Int): Int = {
    if(bust(handA) && bust(handB)){
        0
    } else if(bust(handA)){
        handB
    } else if(bust(handB)){
        handA
    } else {
        maxHand(handA,handB)
    }
}

println(ganador(26,20))
println(ganador(18,22))
println(ganador(24,25))
println(ganador(17,19))
println(ganador(21,20))
```

**Conclusiones y Documentación de Pruebas:**
* `ganador(26, 20)` da **20**: Se cumple la condición del primer `else if` (solamente `handA` supera 21, ya que 26 > 21), por lo que gana directamente `handB` (20).
* `ganador(18, 22)` da **18**: Se cumple la condición del segundo `else if` (solamente `handB` supera 21, ya que 22 > 21), por lo que gana directamente `handA` (18).
* `ganador(24, 25)` da **0**: Se cumple la condición inicial del `if` principal (`bust(handA) && bust(handB)`), ya que ambas manos superan 21, por lo que se devuelve `0` como valor de empate por sobrepasar el límite.
* `ganador(17, 19)` da **19**: Se ejecuta el bloque `else` porque ninguna se pasa de 21; se delega en `maxHand(17, 19)` que determina que 19 es mayor.
* `ganador(21, 20)` da **21**: Se ejecuta el bloque `else` al ser ambas menores o iguales a 21; la función `maxHand(21, 20)` selecciona correctamente el 21.

## Ejercicio 9 — Recorrer un Array con while

**Explicación:** En este ejercicio se recorre un array de enteros que representa diferentes puntuaciones de manos mediante un bucle imperativo while, evaluando cada posición de forma dinámica con una función auxiliar y mostrando el resultado por consola.

```scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0

while(i < manos.length){
    println(manos(i) + " -> "+ bust(manos(i)))
    i += 1
}
```

**Conclusiones y Documentación:**

-   **Por qué el contador debe ser mutable:** El contador `i`necesita ser declarado como `var`porque su valor cambia (se incrementa en 1) en cada iteración del bucle para poder avanzar a la siguiente posición del array.
-   **Qué ocurriría si no incrementaras i:** Si se omite la instrucción `i += 1`, el valor de `i`se mantendría perpetuamente en 0, provocando un bucle infinito que evaluaría una y otra vez el primer elemento hasta agotar los recursos de la máquina o interrumpir la ejecución.
-   **Elementos de estilo imperativo:** El uso de una variable de estado mutable ( `var i = 0`), la estructura de control de flujo basada en bucles condicionales de tipo `while`, y la modificación explícita del índice paso a paso son características puramente imperativas.

## Ejercicio 10 — Operaciones con Listas y Principio de Inmutabilidad

**Explicación:** En este ejercicio se trabaja con colecciones de tipo `List` en Scala. Se inicializa una lista de jugadores, se añade un nuevo elemento al principio utilizando el operador de construcción (`::`), y se evalúan propiedades clave como el formato de cadena (`mkString`), la longitud (`length`) y la inversión de elementos (`reverse`), demostrando la inmutabilidad de las listas.

```scala
val jugadores = List("Alex", "Chen", "Marta")
println(jugadores.mkString(", "))

val jugadores_2 = "Sindhu" :: jugadores
println(jugadores_2.mkString(", "))
println("Longitud: " + jugadores_2.length)
println("Invertida: " + jugadores_2.reverse)

println(jugadores.mkString(", "))
println("Longitud: " + jugadores.length)
println("Invertida: " + jugadores.reverse)
```

**Conclusiones y Documentación:**

-   **El operador de pre-pend (`::` / Cons):** Permite añadir un nuevo elemento al inicio de una lista existente de forma eficiente en la programación funcional.

-   **Inmutabilidad de las listas:** La lista original `jugadores` no sufre ninguna modificación tras la creación de `jugadores_2`, lo que ilustra claramente el principio de inmutabilidad y previene efectos secundarios indeseados.

-   **Métodos nativos de colecciones:** El uso de `mkString` facilita la representación visual limpia de los elementos, mientras que `length` y `reverse` permiten consultar dimensiones y reorganizar colecciones de manera directa.

## Ejercicio 13 foreach y funciones como valores

**Explicación:** Se contrasta el recorrido de colecciones mediante enfoques de control imperativo clásico frente a métodos de orden superior basados ​​en el paradigma funcional.

```scala
val manos = Array(17, 24, 21, 26, 18)

// Versión A: Bucle while imperativo
var i = 0
while (i < manos.length) {
  println(manos(i))
  i += 1
}

// Versión B: Enfoque funcional con foreach
manos.foreach(m => println(m))
```

**Conclusiones y Documentación:**

-   **Necesidad de contadores y variables:** La **Versión A** (bucle `while`) requiere obligatoriamente declarar un contador numérico y una variable mutable (`var i`) para recorrer manualmente el índice del array. Por el contrario, la **Versión B** (`foreach`) encapsula toda la lógica de iteración de forma abstracta.
-   **Estilo funcional:** La **Versión B** se aproxima fielmente al estilo funcional presentado en los materiales del curso, ya que trata la colección como un flujo de datos abstracto al que se le aplica una función anónima sin necesidad de gestionar estados de índices mutables de bajo nivel.

## Ejercicio 14 Efectos secundarios y estilo de programación

**Explicación:** Se analiza el concepto de efectos secundarios ( *side effect* ) observando cómo una función modifica el estado de una variable externa, contraponiéndola con una función pura de estilo funcional.

```scala
var total = 0

def sumarAlTotal(valor: Int) = {
  total = total + valor
}

sumarAlTotal(5)
println(total) 
sumarAlTotal(10)
println(total) 

def sumar(a: Int, b: Int): Int = {
  a + b
}
```

**Conclusiones y Documentación:**

-   **Modificación de ámbito externo:** La función `sumarAlTotal` altera una variable declarada fuera de su bloque de código (`total`), lo cual introduce un **efecto secundario**. Esto hace que el resultado de la función dependa del estado mutable del entorno y no solo de sus parámetros de entrada.

| **Característica** | **sumarAlTotal** | **sumar** |
| --- | --- | --- |
| **Modifica datos externos** | Sí (variable global `total`) | No |
| **Devuelve un resultado calculado** | No (retorna `Unit`) | Sí (retorna el valor operativo `Int`) |
| **Utiliza efecto secundario** | Sí | No (función pura) |
| **Estilo predominante** | Imperativo | Funcional |

## Ejercicio 15 Programa integrado: torneo de Twenty-One

**Explicación:** Programa integral que combina listas inmutables, arrays de soporte, funciones de validación de reglas de juego, estructuras de control y análisis de resultados de múltiples rondas.

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)

def getManosValidas(manosArr: Array[Int]): List[Int] = {
    var manosValidas: List[Int] = List()
    var i = 0

    while(i < manosArr.length) {
        if(bust(manosArr(i))) {
            println("La mano de " + jugadores(i) + " ha superado 21.")
        } else {
            manosValidas = manosArr(i) :: manosValidas
            println("La mano de " + jugadores(i) + " es válida con " + manosArr(i) + ".")
        }
        i += 1
    }
    manosValidas.reverse
}

def getJugadoresValidos(manosArr: Array[Int], jugadoresArr: List[String]): List[String] = {
    var jugadoresValidos: List[String] = List()
    var i = 0

    while(i < manosArr.length) {
        if(!bust(manosArr(i))) {
            jugadoresValidos = jugadoresArr(i) :: jugadoresValidos
        }
        i += 1
    }
    jugadoresValidos.reverse
}

def mostrarGanador(manosArr: List[Int], jugadoresArr: List[String]) = {
    var nombreGanador: String = ""
    var puntuacionGanador: Int = 0
    var i = 0

    if(manosArr.length == jugadoresArr.length && manosArr.length > 0) {
        while(i < manosArr.length) {
            if(manosArr(i) > puntuacionGanador) {
                puntuacionGanador = manosArr(i)
                nombreGanador = jugadoresArr(i)
            }
            i += 1 
        }
        println("El ganador de la partida es " + nombreGanador + " con una puntuación de " + puntuacionGanador)
    } else {
        println("No hay jugadores válidos")
    }
}

// --- Primera Partida ---
println("=== Primera Partida ===")
val manosEnPrimeraPartida: List[Int] = getManosValidas(manos)
val jugadoresEnPrimeraPartida: List[String] = getJugadoresValidos(manos, jugadores)

mostrarGanador(manosEnPrimeraPartida, jugadoresEnPrimeraPartida)

// --- Segunda Partida ---
println("\n=== Segunda Partida ===")
val manosRonda2 = Array(22, 19, 20, 21)

val manosEnSegundaPartida: List[Int] = getManosValidas(manosRonda2)
val jugadoresEnSegundaPartida: List[String] = getJugadoresValidos(manosRonda2, jugadores)

mostrarGanador(manosEnSegundaPartida, jugadoresEnSegundaPartida)
```

**Conclusiones y Documentación:**

-   **Reutilización de código (Principio DRY):** Al encapsular la lógica de filtrado y búsqueda del valor máximo en métodos independientes con parámetros formales, se pueden procesar múltiples escenarios de juego (Ronda 1 y Ronda 2) sin reescribir ni una sola línea de control de flujo.

-   **Independencia de estado entre ejecuciones:** Cada invocación crea sus propias variables locales ( `manosValidas`, `puntuacionGanador`, contadores `i`), garantizando que los resultados acumulados de la primera partida no contaminan ni alteran los cálculos de la segunda.

-   **Robustez ante variaciones posicionales:** En la primera ronda el descarte ocurre en una posición intermedia ( `Chen`con 24), mientras que en la segunda ronda el descarte ocurre en el índice inicial ( `Alex`con 22). En ambos casos, el algoritmo mantiene alineadas las listas resultantes y extrae el ganador exacto ( `Marta`con 21 en la ronda 1 y `Sindhu`con 21 en la ronda 2).
